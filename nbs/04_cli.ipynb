Installing litesearch adds a `litesearch` command. It has one subcommand, `install`, which copies the `SKILL.md` file bundled with the package into a project so that coding agents find it. The file describes litesearch's API for an agent such as Claude Code or Codex. It is written to three places under the target folder: `.agents/skills/litesearch/`, `.claude/skills/litesearch/` and `.Codex/skills/litesearch/`.

```sh
litesearch install                 # into the current git repository's root, or the current folder outside one
litesearch install --dir ~/myapp   # into another folder
```

Running `litesearch` with no subcommand, or an unknown one, prints the list of commands and exits.

````python
#| export
import sys
from fastcore.script import call_parse
from litesearch.data import mv_skill_md
````

````python
#| export
@call_parse
def install(
    dir: str = None  # folder to install into; defaults to the current git repository's root, else the current folder
):
    'Copy the bundled SKILL.md into the .agents, .claude and .Codex skill folders under `dir`.'
    mv_skill_md(dry_run=False, dir=dir)

CMDS = {'install': install}

def main():
    if len(sys.argv) < 2 or sys.argv[1] not in CMDS:
        print('Usage: litesearch <command>')
        print('Commands: ' + ', '.join(CMDS))
        sys.exit(0)
    cmd = sys.argv.pop(1)
    CMDS[cmd]()
````
